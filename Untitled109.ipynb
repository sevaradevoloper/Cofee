```python
import math
import torch
import torch.nn as nn
import torch.nn.functional as F

class SelfAttention(nn.Module):
    def __init__(self, C):
        super().__init__()
        self.q = nn.Linear(C, C, bias=False)
        self.k = nn.Linear(C, C, bias=False)
        self.v = nn.Linear(C, C, bias=False)

    def forward(self, x):
        Q, K, V = self.q(x), self.k(x), self.v(x)
        scores = Q @ K.transpose(-2, -1) / math.sqrt(Q.size(-1))
        w = F.softmax(scores, dim=-1)
        return w @ V, w
```

```python
import torch
import torch.nn as nn


torch.manual_seed(0)
emb = nn.Embedding(3, 8)
sa = SelfAttention(8)

a = emb(torch.tensor([[0, 1, 2]]))
b = emb(torch.tensor([[1, 0, 2]]))

out_a, _ = sa(a)
out_b, _ = sa(b)

print(torch.allclose(out_a[0, 0], out_b[0, 1], atol=0.000001))
print(out_a[0,0])
print(out_b[0,1])
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

torch.manual_seed(0)


class SelfAttention(nn.Module):
    def __init__(self, C):
        super().__init__()
        self.q = nn.Linear(C, C, bias=False)
        self.k = nn.Linear(C, C, bias=False)
        self.v = nn.Linear(C, C, bias=False)

    def forward(self, x):
        B, T, C = x.shape
        q = self.q(x)
        k = self.k(x)
        v = self.v(x)

        # Attention score hisoblash
        scores = q @ k.transpose(-2, -1) / (C ** 0.5)
        w = F.softmax(scores, dim=-1)
        out = w @ v
        return out, w

# --- TOPSHIRIQ KODI ---
emb = nn.Embedding(4, 8)
sa = SelfAttention(8)
ids = torch.tensor([[0, 1, 2, 3]])

# 1-band (Oson)
x = emb(ids)
print("1-band (shape):", x.shape)  # Kutilgan natija: torch.Size([1, 4, 8])

# 2-band (O'rta)
out, _ = sa(x)
out_r, _ = sa(emb(ids.flip(1)))
print("2-band (allclose):", torch.allclose(out.flip(1), out_r, atol=1e-6))

# 3-band (Qiyin)
p = nn.Embedding(4, 8)(torch.arange(4))
out, _ = sa(x + p)
out_r, _ = sa(emb(ids.flip(1)) + p)
print("3-band (position embedding bilan):", torch.allclose(out.flip(1), out_r, atol=1e-6))
```
